```python
import osmnx as ox

# 1. 设置 osmnx 的一些基础配置（可选）
ox.settings.use_cache = True        # 缓存请求，重复运行会更快
ox.settings.log_console = True      # 在控制台打印日志，方便看进度

# 2. 定义研究区域名称（内罗毕）
place_name = "Nairobi, Kenya"

# 3. 下载道路网络
# network_type:
#   'drive'  : 可行驶道路（默认适合交通分析）
#   'all'    : 所有道路，包括小路、步道
#   'walk'   : 行人可走
#   'bike'   : 自行车可走
G = ox.graph_from_place(place_name, network_type="all")

print(G)
```

```python
# 将图转为边（即道路）和点（路口）的 GeoDataFrame
gdf_nodes, gdf_edges = ox.graph_to_gdfs(G)

print(gdf_edges.head())
```

```python
gdf_edges
```

```python
# reset_index(inplace=True, drop=True)  # 重置索引
gdf_edges.reset_index(inplace=True)  # 重置索引
gdf_edges.reset_index(inplace=True, drop=True)  # 重置索引
```

```python
gdf_edges
```

```python
# 处理 highway 字段：如果是列表，只取第一个元素
# Handle cases where 'highway' is a list by taking the first element to fix "unhashable type: 'list'"
gdf_edges['highway'] = gdf_edges['highway'].apply(lambda x: x[0] if isinstance(x, list) else x)
gdf_edges['reversed'] = gdf_edges['reversed'].apply(lambda x: x[0] if isinstance(x, list) else x)
gdf_edges['osmid'] = gdf_edges['osmid'].apply(lambda x: x[0] if isinstance(x, list) else x)
```

```python
# export to gpkg
gdf_edges.to_file("Nairobi_road_edges.gpkg", driver='GPKG')
# export to file if needed
gdf_nodes.to_file("Nairobi_road_nodes.gpkg", driver='GPKG')
```
